```python
# STEP 02: DATA CLEANING

## Objective
#Clean and prepare the E-Commerce dataset for business analysis.

## Tasks
#1. Load datasets
#2. Check duplicate records
#3. Handle missing values
#4. Convert date columns
#5. Validate important columns
#6. Save cleaned datasets

# Tools: Python, Pandas and NumPy





```

```python
#  Cell 2 — Import libraries and create folders

import pandas as pd
import numpy as np
import os

# Create output folder
os.makedirs("processed", exist_ok=True)

print("Libraries imported successfully!")
print("Data cleaning started!")
```

```python
# Cell 3 — Load all datasets

# Load original datasets

orders = pd.read_csv("dataset/olist_orders_dataset.csv")

customers = pd.read_csv("dataset/olist_customers_dataset.csv")

products = pd.read_csv("dataset/olist_products_dataset.csv")

order_items = pd.read_csv("dataset/olist_order_items_dataset.csv")

payments = pd.read_csv("dataset/olist_order_payments_dataset.csv")

reviews = pd.read_csv("dataset/olist_order_reviews_dataset.csv")

sellers = pd.read_csv("dataset/olist_sellers_dataset.csv")

# Store datasets

datasets = {
    "orders": orders,
    "customers": customers,
    "products": products,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "sellers": sellers
}

print("All datasets loaded successfully!")
```

```python
# Cell 4 — Remove duplicate records

# Check and remove exact duplicates

for name in datasets:

    before = len(datasets[name])

    datasets[name] = datasets[name].drop_duplicates()

    after = len(datasets[name])

    print(name)
    print("Duplicates removed:", before - after)
    print("-" * 30)


# Update datasets

orders = datasets["orders"]
customers = datasets["customers"]
products = datasets["products"]
order_items = datasets["order_items"]
payments = datasets["payments"]
reviews = datasets["reviews"]
sellers = datasets["sellers"]

print("Duplicate checking completed!")
```

```python
# Cell 5 — Convert date columns

# Convert date columns into datetime format

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:

    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

print("Date conversion completed!")

display(orders[date_columns].head())
```

```python
# Cell 6 — Missing value analysis

# Check missing values in all datasets

for name, df in datasets.items():

    print("\nDataset:", name)

    missing = df.isnull().sum()

    missing = missing[missing > 0]

    if len(missing) > 0:

        display(missing)

    else:

        print("No missing values found!")
```

```python
#  Cell 7 — Validate important columns

# Check important business columns

print("ORDERS VALIDATION")

print("Missing Order IDs:",
      orders["order_id"].isnull().sum())

print("\nORDER ITEMS VALIDATION")

print("Missing Product IDs:",
      order_items["product_id"].isnull().sum())

print("Missing Prices:",
      order_items["price"].isnull().sum())

print("Missing Freight Values:",
      order_items["freight_value"].isnull().sum())

print("\nPRODUCT VALIDATION")

print("Missing Product IDs:",
      products["product_id"].isnull().sum())
```

```python
# Cell 8 — Save cleaned datasets

# Save cleaned datasets

cleaned_datasets = {
    "orders_clean.csv": orders,
    "customers_clean.csv": customers,
    "products_clean.csv": products,
    "order_items_clean.csv": order_items,
    "payments_clean.csv": payments,
    "reviews_clean.csv": reviews,
    "sellers_clean.csv": sellers
}

for filename, df in cleaned_datasets.items():

    df.to_csv(
        "processed/" + filename,
        index=False
    )

    print("Saved:", filename)

print("\nDATA CLEANING COMPLETED SUCCESSFULLY!")
```
